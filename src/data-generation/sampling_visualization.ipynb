```python
import numpy as np
import matplotlib.pyplot as plt
import os
import json
from matplotlib.patches import Circle, Rectangle

plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern Roman"],
    "axes.labelsize": 12,
    "font.size": 12,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
})
```

```python
data_path = "../../data/generated_data_simple/test"

fig, ax = plt.subplots(1, 1, figsize=(8, 8))

rect = Rectangle((0, 0), 20, 20, linewidth=5, edgecolor='red', facecolor='none')
ax.add_patch(rect)

for fn in os.listdir(data_path):
    params_file = os.path.join(data_path, fn, "params.json")
    with open(params_file, "r") as f:
        params = json.load(f)
    
    x, y, r = 0, 0, 0
    for g in params["geometry"]:
        if g["type"] == "cylinder":
            x = g["pos"]["x"]
            y = g["pos"]["y"]
            r = g["size"]
            break

    ax.scatter(x, y, color="black")

    circle = Circle((x, y), r, edgecolor="black", facecolor="none", linewidth=1)
    ax.add_patch(circle)

ax.set_xlabel("Cylinder x")
ax.set_ylabel("Cylinder y")
plt.show()
```

```python

for fn in os.listdir(data_path):
    u_star = np.load(os.path.join(data_path, fn, "u_star.npy"))
    
    unique_u_star = []

    for u in u_star:
        remove_u = False
        for u_ in unique_u_star:
            diff = np.mean((u - u_) ** 2) / np.mean(u_ ** 2)
        
            if diff < 1e-5:
                remove_u = True
                break
            
        if not remove_u:
            unique_u_star.append(u)

    print(len(u_star), len(unique_u_star))


    break
```
